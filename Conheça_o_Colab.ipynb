```
# ============================================================
# 📘 TRABALHO: Chatbots Sem Alucinações - UNIFECAF AI
# Aluna: Joelma dos Santos Ramos
# Disciplina: Gen AI and Prompt Engineering
# Professor: Rodrigo Moreira
# Data: 09/11/2025
# ============================================================

# 1️⃣ Instalar biblioteca necessária
!pip install reportlab

# 2️⃣ Importar bibliotecas
from reportlab.lib.pagesizes import A4
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, PageBreak
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib.enums import TA_CENTER
from reportlab.pdfbase.cidfonts import UnicodeCIDFont
from reportlab.pdfbase import pdfmetrics

# 3️⃣ Registrar fonte compatível com acentos
pdfmetrics.registerFont(UnicodeCIDFont('HeiseiMin-W3'))

# 4️⃣ Criar o documento PDF
file_path = "/content/Chatbots_Sem_Alucinacoes_UNIFECAF_AI_JoelmaRamos.pdf"
pdf = SimpleDocTemplate(file_path, pagesize=A4)
styles = getSampleStyleSheet()
styles.add(ParagraphStyle(name='CenterTitle', alignment=TA_CENTER, fontName='HeiseiMin-W3', fontSize=14, leading=18))
styles.add(ParagraphStyle(name='BodyText2', fontName='HeiseiMin-W3', fontSize=12, leading=16))

# 5️⃣ Capa
elements = []
elements.append(Spacer(1, 100))
elements.append(Paragraph("CENTRO UNIVERSITÁRIO UNIFECAF", styles['CenterTitle']))
elements.append(Spacer(1, 20))
elements.append(Paragraph("Disciplina: Gen AI and Prompt Engineering", styles['CenterTitle']))
elements.append(Spacer(1, 20))
elements.append(Paragraph("Professor: Rodrigo Moreira", styles['CenterTitle']))
elements.append(Spacer(1, 100))
elements.append(Paragraph("Chatbots Sem Alucinações:", styles['CenterTitle']))
elements.append(Paragraph("O Desafio da Startup UNIFECAF AI para Garantir Precisão e Confiabilidade", styles['CenterTitle']))
elements.append(Spacer(1, 150))
elements.append(Paragraph("Aluna: Joelma dos Santos Ramos", styles['CenterTitle']))
elements.append(Paragraph("Data: 09 de novembro de 2025", styles['CenterTitle']))
elements.append(PageBreak())

# 6️⃣ Introdução
elements.append(Paragraph("<b>Introdução</b>", styles['Heading2']))
elements.append(Paragraph("""
A Startup UNIFECAF AI foi criada em 2024 com o objetivo de desenvolver soluções inovadoras de Inteligência Artificial aplicadas ao ambiente acadêmico.
Seu primeiro grande projeto foi um chatbot com IA generativa, desenvolvido para auxiliar estudantes em dúvidas sobre matrícula, prazos e serviços institucionais.
Contudo, o aumento do uso revelou um problema crítico: o chatbot começou a apresentar alucinações, fornecendo informações incorretas, desatualizadas e até inventadas,
comprometendo a credibilidade da instituição. Este trabalho busca analisar o problema e propor soluções para garantir a precisão e a confiabilidade das respostas geradas.
""", styles['BodyText2']))
elements.append(Spacer(1, 12))

# 7️⃣ Desenvolvimento
elements.append(Paragraph("<b>Desenvolvimento</b>", styles['Heading2']))
elements.append(Paragraph("""
As chamadas "alucinações" em modelos de IA generativa ocorrem quando o sistema cria respostas que parecem corretas, mas não se baseiam em fatos reais.
No caso da UNIFECAF AI, as respostas incorretas incluíam datas de matrícula erradas e valores de mensalidades desatualizados.
Esse problema é amplamente conhecido no mercado e várias empresas têm desenvolvido estratégias para reduzir tais falhas.

Entre as principais soluções estão as técnicas de RAG (Retrieval-Augmented Generation), que combinam geração de texto com busca de informações reais em bases de dados verificadas,
e o fine-tuning com dados institucionais, que garante que o modelo utilize informações específicas e atualizadas da organização.
Outras medidas incluem monitoramento humano e filtros automáticos de verificação de fatos.

Um exemplo concreto é o Microsoft Copilot for Education, que utiliza RAG para acessar dados oficiais da instituição, garantindo respostas precisas e seguras.
""", styles['BodyText2']))
elements.append(Spacer(1, 12))

# 8️⃣ Simulação no Colab
elements.append(Paragraph("<b>Simulação no Google Colab</b>", styles['Heading2']))
elements.append(Paragraph("""
Para demonstrar uma abordagem prática, foi desenvolvido um código simples em Python que simula um chatbot institucional confiável.
A solução utiliza busca semântica em uma base de dados verificada, evitando a geração de respostas inventadas.
Abaixo segue o exemplo de código utilizado no Google Colab:
""", styles['BodyText2']))
elements.append(Spacer(1, 12))
elements.append(Paragraph("""
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

base_conhecimento = {
    "matricula": "As matrículas da UniFECAF ocorrem entre 10 e 20 de janeiro de 2025.",
    "mensalidade": "O valor da mensalidade de 2025 é de R$ 890,00 para cursos presenciais.",
    "documentos": "Os documentos exigidos são RG, CPF, comprovante de residência e histórico escolar.",
    "calendario": "O calendário acadêmico está disponível no site oficial: www.unifecaf.edu.br."
}

def buscar_resposta(pergunta):
    perguntas = list(base_conhecimento.keys())
    respostas = list(base_conhecimento.values())
    vectorizer = TfidfVectorizer()
    matriz = vectorizer.fit_transform(perguntas + [pergunta])
    similaridades = cosine_similarity(matriz[-1], matriz[:-1])
    indice = similaridades.argmax()
    return respostas[indice]
""", styles['BodyText2']))
elements.append(Spacer(1, 12))

# 9️⃣ Conclusão
elements.append(Paragraph("<b>Conclusão</b>", styles['Heading2']))
elements.append(Paragraph("""
O desafio enfrentado pela UNIFECAF AI reflete um dos maiores obstáculos da IA generativa moderna: a confiabilidade das respostas.
Para evitar alucinações, é essencial implementar soluções técnicas como RAG, bases de dados verificadas e revisões humanas periódicas.
Com essas medidas, o chatbot poderá fornecer informações precisas e fortalecer a credibilidade da instituição, tornando-se uma ferramenta eficiente e segura no ambiente acadêmico.
""", styles['BodyText2']))

# 🔟 Gerar o PDF
pdf.build(elements)
print(f"✅ PDF gerado com sucesso: {file_path}")
```
