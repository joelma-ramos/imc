```
# Instalar biblioteca necessária
!pip install reportlab

import os
import matplotlib.pyplot as plt
import numpy as np
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle

# --- Criar a pasta /mnt/data caso ela não exista ---
if not os.path.exists("/mnt/data"):
    os.makedirs("/mnt/data")

# Caminhos dos gráficos
chart1_path = "/mnt/data/chart1.png"
chart2_path = "/mnt/data/chart2.png"

# ----------- GERAR GRÁFICOS -----------
# Gráfico 1
plt.figure()
x = np.arange(5)
y = np.random.randint(10, 50, size=5)
plt.plot(x, y, marker='o')
plt.title("Volume de Feedbacks por Categoria")
plt.savefig(chart1_path)
plt.close()

# Gráfico 2
plt.figure()
labels = ['Positivo', 'Neutro', 'Negativo']
sizes = [60, 25, 15]
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title("Distribuição de Sentimentos")
plt.savefig(chart2_path)
plt.close()

# ----------- GERAR O PDF -----------
pdf_path = "/mnt/data/Relatorio_Joelma.pdf"

styles = getSampleStyleSheet()
styles.add(ParagraphStyle(name="Titulo", fontSize=18, alignment=1, spaceAfter=20))
styles.add(ParagraphStyle(name="Corpo", fontSize=11, leading=16))

doc = SimpleDocTemplate(pdf_path, pagesize=A4)
content = []

content.append(Paragraph("RELATÓRIO ACADÊMICO – NLP", styles["Titulo"]))
content.append(Paragraph("Aluna: JOELMA DOS SANTOS RAMOS", styles["Corpo"]))
content.append(Spacer(1, 20))

content.append(Paragraph("<b>Gráfico 1 – Volume de Feedbacks</b>", styles["Corpo"]))
content.append(Image(chart1_path, width=400, height=300))
content.append(Spacer(1, 20))

content.append(Paragraph("<b>Gráfico 2 – Distribuição de Sentimentos</b>", styles["Corpo"]))
content.append(Image(chart2_path, width=400, height=300))
content.append(Spacer(1, 20))

doc.build(content)

pdf_path
```
