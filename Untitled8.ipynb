```python
import pandas as pd
import numpy as np
import nltk
import re
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import accuracy_score, f1_score, classification_report
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import LinearSVC
from xgboost import XGBClassifier
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from scikeras.wrappers import KerasClassifier
from nltk.corpus import stopwords

# Baixar stopwords
nltk.download("stopwords")
stop_words = stopwords.words("portuguese")
```

```python
# Exemplo de dataset fictício de chamados
data = {
    "mensagem": [
        "Meu pedido não chegou e já faz mais de uma semana!",
        "Gostaria de saber o prazo de entrega, por favor.",
        "A embalagem veio danificada, preciso de reembolso urgente.",
        "Como faço para alterar meu endereço de entrega?",
        "O produto não liga, quero trocar imediatamente!",
        "A entrega foi perfeita, obrigado pelo atendimento.",
        "Tenho dúvidas sobre o pagamento via boleto.",
        "Não consigo acessar minha conta, preciso de ajuda agora!",
        "O chat está demorando muito, isso é urgente!",
        "Quero apenas confirmar o status do meu pedido."
    ],
    "urgencia": [
        "Alta", "Baixa", "Alta", "Média", "Alta",
        "Baixa", "Média", "Alta", "Alta", "Baixa"
    ]
}

df = pd.DataFrame(data)
```

```python
def limpar_texto(texto):
    texto = texto.lower()
    texto = re.sub(r"[^a-zA-Zá-úÁ-Ú\s]", "", texto)
    palavras = texto.split()
    palavras = [p for p in palavras if p not in stop_words]
    return " ".join(palavras)

df["texto_limpo"] = df["mensagem"].apply(limpar_texto)

# Vetorização TF-IDF
tfidf = TfidfVectorizer(max_features=100)
X = tfidf.fit_transform(df["texto_limpo"]).toarray()

# Codificação das classes
le = LabelEncoder()
y = le.fit_transform(df["urgencia"])

# Divisão treino/teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Random Forest
rf = RandomForestClassifier(random_state=42)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)

# SVM
svm = LinearSVC()
svm.fit(X_train, y_train)
svm_pred = svm.predict(X_test)

# XGBoost
xgb = XGBClassifier(use_label_encoder=False, eval_metric="mlogloss")
xgb.fit(X_train, y_train)
xgb_pred = xgb.predict(X_test)
```

```python
def criar_mlp():
    model = Sequential()
    model.add(Dense(64, activation='relu', input_dim=X_train.shape[1]))
    model.add(Dropout(0.3))
    model.add(Dense(32, activation='relu'))
    model.add(Dropout(0.3))
    model.add(Dense(3, activation='softmax'))
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
    return model

mlp = KerasClassifier(model=criar_mlp, epochs=10, batch_size=4, verbose=0)
mlp.fit(X_train, y_train)
mlp_pred = mlp.predict(X_test)
```

```python
print(classification_report(
    y_test,
    xgb_pred,
    labels=np.arange(len(le.classes_)),
    target_names=le.classes_
))
```
