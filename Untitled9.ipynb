```python
# === CÉLULA 2 — Imports e configurações iniciais ===
import os
import random
import numpy as np
import pandas as pd
import re
import string
import joblib
from datetime import datetime

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import LinearSVC
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report, confusion_matrix

import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords

from scikeras.wrappers import KerasClassifier
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)
```

```python
# === CÉLULA 3 — Carregar dataset (substitua o caminho pelo seu CSV) ===
# O código tenta abrir "/content/calls.csv". Se não existir, usa um dataset fictício de exemplo.
DATA_PATH = "/content/calls.csv"

if os.path.exists(DATA_PATH):
    df = pd.read_csv(DATA_PATH)
    print("Dataset carregado:", DATA_PATH)
else:
    print("Arquivo not found:", DATA_PATH, "-- usando dataset de exemplo.")
    data = {
        "message": [
            "Meu pedido não chegou, preciso urgente",
            "Gostaria de trocar o endereço de entrega",
            "A cobrança veio errada, quero reembolso",
            "Como atualizo minha senha?",
            "Erro no pagamento, cartão recusado",
            "A entrega foi ótima, obrigado",
            "Não consigo acessar a conta, socorro",
            "Informação sobre prazo de entrega",
            "Produto com defeito, quero troca",
            "Consulta simples sobre nota fiscal"
        ],
        "urgency": [
            "Alta", "Baixa", "Alta", "Baixa", "Alta",
            "Baixa", "Alta", "Baixa", "Média", "Baixa"
        ]
    }
    df = pd.DataFrame(data)

# Ajuste estes nomes se seu CSV tem colunas diferentes
TEXT_COL = "message"
LABEL_COL = "urgency"

print("Shape:", df.shape)
display(df.head())
print("Distribuição de classes:\n", df[LABEL_COL].value_counts())
```

```python
# === CÉLULA 4 — Pré-processamento de texto e TF-IDF ===
stop_words = set(stopwords.words("portuguese"))

def clean_text(text):
    if pd.isna(text):
        return ""
    text = str(text).lower()
    text = re.sub(r"http\S+|www\.\S+", " ", text)                # urls
    text = re.sub(r"\S+@\S+", " ", text)                        # emails
    text = re.sub(r"[^0-9a-zà-ú\s]", " ", text)                 # remove pontuação (mantém acentos)
    tokens = text.split()
    tokens = [t for t in tokens if t not in stop_words]
    return " ".join(tokens)

df["text_clean"] = df[TEXT_COL].apply(clean_text)

# Vetorização TF-IDF
tfidf = TfidfVectorizer(ngram_range=(1,2), min_df=1, max_features=5000)
X = tfidf.fit_transform(df["text_clean"])
# codificar labels
le = LabelEncoder()
y = le.fit_transform(df[LABEL_COL])

print("TF-IDF shape:", X.shape)
print("Classes:", list(le.classes_))
```

```python
import pandas as pd

print(pd.Series(y).value_counts())
```

```python
mask = y != 'Baixa'
X = X[mask]
y = y[mask]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)
```

```python
from sklearn.utils import resample
import pandas as pd

# Juntar X e y temporariamente
df = pd.DataFrame(X)
df['label'] = y

# Descobrir a classe minoritária
classe_minoritaria = df['label'].value_counts().idxmin()

# Fazer oversampling (duplicar)
df_min = df[df['label'] == classe_minoritaria]
df_extra = resample(df_min, replace=True, n_samples=2, random_state=42)

# Concatenar novamente
df_bal = pd.concat([df, df_extra])

# Separar X e y novamente
X = df_bal.drop('label', axis=1)
y = df_bal['label']
```

```python
# === CÉLULA 6 — Função utilitária de avaliação ===
def evaluate_and_report(name, model, Xtr, Xte, ytr, yte, label_encoder):
    # se o modelo exigir array denso, converter
    try:
        y_pred = model.predict(Xte)
    except Exception:
        y_pred = model.predict(Xte.toarray())
    acc = accuracy_score(yte, y_pred)
    f1_macro = f1_score(yte, y_pred, average="macro")
    f1_weighted = f1_score(yte, y_pred, average="weighted")
    # Garantir que classification_report não quebre: especificar labels
    labels = np.arange(len(label_encoder.classes_))
    report = classification_report(yte, y_pred, labels=labels, target_names=label_encoder.classes_, zero_division=0)
    print(f"--- {name} ---")
    print("Accuracy:", round(acc, 4))
    print("F1 macro:", round(f1_macro, 4), "F1 weighted:", round(f1_weighted, 4))
    print(report)
    cm = confusion_matrix(yte, y_pred, labels=labels)
    return {"model": name, "accuracy": acc, "f1_macro": f1_macro, "f1_weighted": f1_weighted, "cm": cm}
```

```python
# === CÉLULA 7 — Treinar RandomForest, SVM e XGBoost ===
results = []

# Random Forest
rf = RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE, class_weight="balanced")
rf.fit(X_train, y_train)
results.append(evaluate_and_report("RandomForest", rf, X_train, X_test, y_train, y_test, le))

# Linear SVM (usa vetores esparsos)
svc = LinearSVC(C=1.0, class_weight="balanced", random_state=RANDOM_STATE, max_iter=10000)
svc.fit(X_train, y_train)
results.append(evaluate_and_report("LinearSVC", svc, X_train, X_test, y_train, y_test, le))

# XGBoost
xgb = XGBClassifier(use_label_encoder=False, eval_metric="mlogloss", random_state=RANDOM_STATE)
# XGBoost aceita matrizes densas or csr; passamos X_train directly
xgb.fit(X_train, y_train)
results.append(evaluate_and_report("XGBoost", xgb, X_train, X_test, y_train, y_test, le))
```

```python
# === CÉLULA 8 — MLP (Keras via scikeras) ===
# scikeras espera arrays densos
X_train_arr = X_train.toarray()
X_test_arr = X_test.toarray()

def create_mlp_model():
    model = Sequential()
    model.add(Dense(128, activation="relu", input_shape=(X_train_arr.shape[1],)))
    model.add(Dropout(0.3))
    model.add(Dense(64, activation="relu"))
    model.add(Dropout(0.2))
    model.add(Dense(len(le.classes_), activation="softmax"))
    model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
    return model

mlp = KerasClassifier(model=create_mlp_model, epochs=12, batch_size=16, verbose=0, random_state=RANDOM_STATE)
mlp.fit(X_train_arr, y_train)
# avaliar
y_pred_mlp = mlp.predict(X_test_arr)
acc = accuracy_score(y_test, y_pred_mlp)
f1_macro = f1_score(y_test, y_pred_mlp, average="macro")
f1_weighted = f1_score(y_test, y_pred_mlp, average="weighted")
print("--- MLP (Keras) ---")
print("Accuracy:", round(acc,4))
print("F1 macro:", round(f1_macro,4), "F1 weighted:", round(f1_weighted,4))
print(classification_report(y_test, y_pred_mlp, labels=np.arange(len(le.classes_)), target_names=le.classes_, zero_division=0))
results.append({"model":"MLP", "accuracy":acc, "f1_macro":f1_macro, "f1_weighted":f1_weighted, "cm": confusion_matrix(y_test, y_pred_mlp, labels=np.arange(len(le.classes_)))})
```

```python
# === CÉLULA 9 — Resumo comparativo e salvar métricas ===
df_results = pd.DataFrame([{"Modelo": r["model"], "Accuracy": r["accuracy"], "F1_macro": r["f1_macro"], "F1_weighted": r["f1_weighted"]} if isinstance(r, dict) and "model" in r else {"Modelo": r["model"], "Accuracy": r["accuracy"], "F1_macro": r["f1_macro"], "F1_weighted": r["f1_weighted"]} for r in results])
# Se algum item for formato diferente (como último), ajustar:
if df_results.empty:
    # fallback: build from results list
    rows = []
    for r in results:
        rows.append({"Modelo": r.get("model", r.get("Modelo")), "Accuracy": r.get("accuracy"), "F1_macro": r.get("f1_macro"), "F1_weighted": r.get("f1_weighted")})
    df_results = pd.DataFrame(rows)

# Ordenar por F1_macro
df_results = df_results.sort_values(by="F1_macro", ascending=False).reset_index(drop=True)
print("\n=== Tabela comparativa ===")
print(df_results)

# salvar CSV com métricas
out_metrics_path = "/content/metrics_results.csv"
df_results.to_csv(out_metrics_path, index=False)
print("Métricas salvas em:", out_metrics_path)
```

```python
# === CÉLULA 10 — Salvar artefatos (vetorizador + melhor modelo como exemplo) ===
# escolher melhor modelo pela F1_macro
best_model_name = df_results.loc[0, "Modelo"]
print("Melhor modelo:", best_model_name)

# salvar tfidf
joblib.dump(tfidf, "/content/tfidf_vectorizer.joblib")
# salvar modelos
joblib.dump(rf, "/content/model_rf.joblib")
joblib.dump(svc, "/content/model_svc.joblib")
joblib.dump(xgb, "/content/model_xgb.joblib")
# Para o MLP (scikeras) salvar com joblib também
joblib.dump(mlp, "/content/model_mlp_scikeras.joblib")

print("Artefatos salvos em /content/: tfidf_vectorizer.joblib, model_*.joblib")
```
